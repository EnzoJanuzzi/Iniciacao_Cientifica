# Criação das Nanopartículas e Intermediários

Utilização do módulo ASE para criar as moléculas, salvando as coordenadas em arquivo POSCAR para ser usado no VASP.

## Importações:

```python
import os

from ase.cluster import Octahedron, Icosahedron
from ase import Atoms
from ase.build import bulk, molecule
from ase.visualize import view
from ase.io import write, read
from ase.collections import g2
```

## NPs:

### NP Ru:

```python
NP_Ru = Icosahedron(symbol='Ru', noshells=3, latticeconstant=3.61)
atoms_ru = NP_Ru.get_global_number_of_atoms()
print(atoms_ru)
filename = 'NP_Ru' + str(atoms_ru) + '.xyz'
write(filename, NP_Ru)
view(NP_Ru, viewer='x3d')
```

```python
CONTCAR_Ru = read('CONTCAR_Ru', format='vasp')
view(CONTCAR_Ru, viewer='x3d')
```

```python
# Setting the cell size:
NP_Ru.set_cell([[20.0, 0.00, 0.00], [0.0, 20.00, 0.00], [0.0, 0.00, 20.00]])
# Translating the nanoparticle to the center of the cell:
NP_Ru.translate([10.0, 10.0, 10.0])
# Creating a POSCAR file for VASP calculations, sorting the coordinates:
write('POSCAR_Ru' + str(atoms_ru), NP_Ru, format='vasp', sort=True)
```

### NP Ru@Cu HCP-shell:

```python
NP_RuCu_HCP = Icosahedron(symbol='Ru', noshells=3, latticeconstant=3.61)
atoms_rucu_hcp = NP_RuCu_HCP.get_global_number_of_atoms()
print(atoms_rucu_hcp)
#filename = 'NP_RuCu_FCC' + str(atoms_ru) + '.xyz'
#write(filename, NP_RuCu_FCC)
view(NP_RuCu_HCP, viewer='x3d')
```

```python
i = 0
core = 0
shell = 0

for vector in NP_RuCu_HCP.positions:
    radius = (vector[0]**2 + vector[1]**2 + vector[2]**2)**0.5
    if radius < 3.2:
        print(i, 'Core', radius)
        NP_RuCu_HCP.numbers[i] = 29 # Changing Ru to Cu
        core += 1
    else:
        print(i, 'Shell', radius)
        shell += 1
    i += 1
```

```python
print('Number of core atoms:', core)
print('Number of shell atoms:', shell)
```

```python
view(NP_RuCu_HCP, viewer='x3d')
```

```python
# Setting the cell size:
NP_RuCu_HCP.set_cell([[20.0, 0.00, 0.00], [0.0, 20.00, 0.00], [0.0, 0.00, 20.00]])
# Translating the nanoparticle to the center of the cell:
NP_RuCu_HCP.translate([10.0, 10.0, 10.0])
# Creating a POSCAR file for VASP calculations, sorting the coordinates:
write('POSCAR_RuCu_HCP', NP_RuCu_HCP, format='vasp', sort=True)
```
